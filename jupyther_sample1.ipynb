```python
def Weather_Intensity():
    temperature = int(input("What is the current temperature? "))
    if temp < 0:
        print("It's freezing cold!")
    elif 1 < temp < 15:
        print("It's cold!")
    elif 16 <= temp < 30:
        print("It's hot!")
    elif 31 < temp <= 45:
        print("It's really hot!")
    else:
        print("It's extremely hot!")
```

```python
Weather_Intensity()
```
